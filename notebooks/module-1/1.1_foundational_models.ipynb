## Agent初始化化和调用

#### 刚才谈的是基本的大模型，现在我们做一个简单的 agent，agent 是围绕大模型做更多的定制化，包括对话的角色，对话历史, 工具的调用, memory 等等. agent 可以大大提高模型的适用度， 也是目前一个行业的发展趋势。

#### 我们使用基于langgraph封装的agent 功能。langgraph 是langchain针对企业级的agent 应用框架，特别适用于企业级安全可控、可管理的智能体的开发。

````python
from langchain.agents import create_agent

agent = create_agent(model=model)
````

````python
agent = create_agent(model="ollama:qwen2.5:14b")
````

````python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="月球的首都是哪里?")]}
)
````

#### 可以看到·基于agent 那它的对话，会有清晰的不同的角色定义，包括人、AI 以及未来的工具

````python
from pprint import pprint

pprint(response)
````

````python
print(response['messages'][-1].content)
````

#### 我们也可以先定好一系列的对话， 发给 agent，请他做最后的回答。

````python
from langchain.messages import AIMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City")]}
)

pprint(response)
````

## 流式输出

#### agent 的对话有一个问题，就是可能会存在延迟，例如你问了一个比较复杂的问题，agent 会计算, 搜索网络，可能经过一段时间才能给你回，那为了提高体验，我们可以把它变成流式输出啊， 伴随着信息的生成，把内容逐步显示出来。


````python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="告诉我关于月球之都月城的一切")]},
    stream_mode="messages"
):

    # token 是包含 token 内容的消息块
    # metadata 包含产生该 token 的节点信息
    
    if token.content:  # 检查是否有实际内容
        print(token.content, end="", flush=True)  # 打印 token
````
